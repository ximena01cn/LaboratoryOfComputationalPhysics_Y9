1\. Write a function that converts number representation, bin<->dec<->hex. (Clearly using the corresponding python built-in functions is not fair..)

```python
# binary to dec
def bin2dec(b):
    dec = 0
    for bit in b:
        dec = dec * 2 + int(bit)
    return dec


def dec2bin(n):
    if n == 0:
        return "0"

    b = ""
    while n > 0:
        b = str(n % 2) + b
        n = n // 2
    return b


def dec2hex(n):
    if n == 0:
        return "0"

    digits = "0123456789ABCDEF"
    h = ""

    while n > 0:
        r = n % 16
        h = digits[r] + h
        n = n // 16

    return h


def hex2dec(h):
    digits = "0123456789ABCDEF"
    dec = 0

    for digit in h.upper():
        dec = dec * 16 + digits.index(digit)

    return dec


def convert(number, base_from, base_to):
    if base_from == "bin":
        dec = bin2dec(number)
    elif base_from == "dec":
        dec = int(number)
    elif base_from == "hex":
        dec = hex2dec(number)

    if base_to == "bin":
        return dec2bin(dec)
    elif base_to == "dec":
        return str(dec)
    elif base_to == "hex":
        return dec2hex(dec)
    
```

3\. Write a program to determine the underflow and overflow limits (within a factor of 2) for python on your computer. 

**Tips**: define two variables inizialized to 1 and halve/double them enough time to exceed the under/over-flow limits  

```python
# Overflow
x = 1.0
# Multiplicar por dos hasta llevar al limite que en este caso seria inf en python
while x * 2 != float("inf"):
    x = x * 2
print("Overflow limit:", x)

# Underflow
y = 1.0
# dividir entre dos hasta llegar lo mas cercano a cero, una siguietne division haria que el resultado fuera 0
while y / 2 != 0:
    y = y / 2
print("Underflow limit:", y)
```

```python
#c

import math
import cmath

def quadratic(a, b, c):
    D = b**2 - 4*a*c

    if D >= 0:
        sqrtD = math.sqrt(D)
    else:
        sqrtD = cmath.sqrt(D)

    if b >= 0:
        q = -0.5 * (b + sqrtD)
    else:
        q = -0.5 * (b - sqrtD)

    if q == 0:
        x1 = 0
        x2 = 0
    else:
        x1 = q / a
        x2 = c / q

    print("First root:", x1)
    print("Second root:", x2)


quadratic(0.001, 1000, 0.001)
```
