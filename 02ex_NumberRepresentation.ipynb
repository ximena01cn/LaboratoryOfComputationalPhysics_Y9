1\. Write a function that converts number representation, bin<->dec<->hex. (Clearly using the corresponding python built-in functions is not fair..)

```python
# binario a decimal, recibe un numero binario en texto
def bin2dec(b):
    dec = 0
    i = len(b) - 1

    for bit in b:
        dec += int(bit) * 2**i # aplicamos formula de multiplicar el bit por dos elevado a la posicion en la cadena
        i -= 1

    return dec               


def dec2bin(n):                
    if n == 0:                 
        return "0"             

    b = ""                      # Cadena vacía para construir el binario

    while n > 0:                # Repite el proceso mientras n sea positivo
        b = str(n % 2) + b      # Obtiene el residuo entre 0 y 1 y lo añade al inicio
        n = n // 2              # Divide n entre 2 y conserva solo la parte entera

    return b                    


def dec2hex(n): # similar a binario pero con las letras de hex
    if n == 0:
        return "0"

    digits = "0123456789ABCDEF"
    h = ""

    while n > 0:
        r = n % 16 #tomamos modulo
        h = digits[r] + h # buscamos a que hex num corresponde y lo agregamos a h, lo añade al inicio
        n = n // 16 # dividimos netre 16 y asi hasta que ya no se pueda

    return h


def hex2dec(h):
    digits = "0123456789ABCDEF"  # Simbolos hex
    dec = 0                      # resultado decimal
    i = len(h) - 1               # exponente primera posicion

    for digit in h.upper():      # Recorre los dígitos de izquierda a derecha.
        value = digits.index(digit)  # Valor numérico del dígito, es decir, l aposicion
        dec += value * 16**i     # Multiplica el valor por su potencia de 16
        i -= 1                   # Disminuye el exponente para el siguiente dígito

    return dec                   


# aqui solo juntamos las funciones implementadas 
# toma culaquier base de entrada, lo convierte a decimal, y de decimal lo conierte a cualquier otra represetnacion
# dado que esas fueron las funciones implementadas
def convert(number, base_from, base_to):
    # de binario a decimal
    if base_from == "bin":
        dec = bin2dec(number)
    # de dec a dec
    elif base_from == "dec":
        dec = int(number)
    # de hex a dec
    elif base_from == "hex":
        dec = hex2dec(number)

    # toma lo que se genera en el anterior, todo lo anteior entra en dec

    # lo de dec sale a bin
    if base_to == "bin":
        return dec2bin(dec)
    # lo de dec sale a dec
    elif base_to == "dec":
        return str(dec)
    # lo de dec sale a hex
    elif base_to == "hex":
        return dec2hex(dec)


convert("FE501", "hex", "bin")
    
```

3\. Write a program to determine the underflow and overflow limits (within a factor of 2) for python on your computer. 

**Tips**: define two variables inizialized to 1 and halve/double them enough time to exceed the under/over-flow limits  

```python
# Overflow
x = 1.0
# Multiplicar por dos hasta llevar al limite que en este caso seria inf en python
while x * 2 != float("inf"):
    x = x * 2
print("Overflow limit:", x)

# Underflow
y = 1.0
# dividir entre dos hasta llegar lo mas cercano a cero, una siguietne division haria que el resultado fuera 0
while y / 2 != 0:
    y = y / 2
print("Underflow limit:", y)
```

7\. Consider the integral of the semicircle of radius 1:
$$
I=\int_{-1}^{1} \sqrt(1-x^2) {\rm d}x
$$
which it's known to be $I=\frac{\pi}{2}=1.57079632679...$.
Alternatively we can use the Riemann definition of the integral:
$$
I=\lim_{N\to\infty} \sum_{k=1}^{N} h y_k 
$$

with $h=2/N$ the width of each of the $N$ slices the domain is divided into and where
$y_k$ is the value of the function at the $k-$th slice.

(a) Write a programe to compute the integral with $N=100$. How does the result compares to the true value?

(b) How much can $N$ be increased if the computation needs to be run in less than a second? What is the gain in running it for 1 minute? 


```python
# a)
def f(x):
    return np.sqrt(1-x**2)
def intt(f, N):
    itn = 0
    # calculamos le ancho de cada intervalo, en este caso es [-1, 1] 1 - (-1) = 2
    h = 2 / N
    for k in range(1, N+1):
        # Empezamos en -1 y avanzamos k veces el ancho h, lo que hace que temrinos en 1 (se cubre todo el intervalo)
        x = -1 + k * h 
        itn += h * f(x)
    return float(itn)

res = intt(f,100)

print("Valor obtenido:", res,".Valor esperado:", np.pi/2)
print("Diferencia entre valor obtenido y valor esperado:", abs(res - np.pi / 2))
```

```python
# b)
import time

def one_sec():
    N = 100
    start = time.time()

    while True:
        result = intt(f, N)
        elapsed = time.time() - start

        # Si el cálculo tarda 1 segundo o más, paramos.
        if elapsed >= 1:
            break

        # Si todavía tarda menos de 1 segundo,
        # aumentamos N para hacer más intervalos.
        N *= 2

    print("N maxima completada en menos de 1s:", N // 2)
    print("Approximation:", result, "Differnece between real value:", abs(result - np.pi / 2))
    print("Elapsed time:", elapsed, "seconds")

one_sec()
```
