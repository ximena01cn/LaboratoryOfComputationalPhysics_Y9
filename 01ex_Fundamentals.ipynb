2\. Convert the following function into a pure function with no global variables or side effects

````python
x = 5
def f(alist):
    for i in range(x):
         alist.append(i)
    return alist

alist = [1,2,3]
ans = f(alist)
print (ans)
print (alist) # alist has been changed!
````

3\. Write a `decorator` hello that makes every wrapped function print “Hello!”, i.e. something like:

```python
@hello
def square(x):
    return x*x
```



````python
def hello(func):
    def wrapper(x):
        a = func(x)
        print("Hello!")
        return a # regresa resultado de a
    return wrapper # imprime resultado de wrapper que es un Hello!

# esto simplemente hace un square = hello(square)
@hello
def square(x):
    return x*x
````

````python
a = square(6)
print(a)
````

4\. Write the factorial function so that it a) does and b) does not use recursion.

````python
# a) with recursion
def fact(n):
    if n == 0:
        return 1
    else:
        n = n * fact(n-1)
    return n

fact(3)
````

````python
#b) with no recursion
def fact_no_r(n):
    factt = 1
    mult = [n]
    while n != 1:
        n = n - 1
        mult.append(n)

    for num in mult:
        # similar a +=, en donde factt se multiplica por num, factt ahora es el resultado de la multiplicacion
        # se actualiza con cada multiplicación
        factt *= num   
    return factt
fact_no_r(3)
````

5\. Use HOFs (zip in particular) to compute the weight of a circle, a disk and a sphere, assuming different radii and different densities:

```python
densities = {"Al":[0.5,1,2],"Fe":[3,4,5],"Pb": [15,20,30]}
radii = [1,2,3]
```

where the entries of the dictionary's values are the linear, superficial and volumetric densities of the materials respectively.

In particular define a list of three lambda functions using a comprehension that computes the circumference, the area and the volume for a given radius.


````python
import numpy as np
densities = {"Al":[0.5,1,2],"Fe":[3,4,5],"Pb": [15,20,30]}
radii = [1,2,3]
````

````python
# Se definen las funciones geométricas mediante una comprensión
geometries = [(2*np.pi, 1), (np.pi, 2), (4/3*np.pi, 3)]

functions = [lambda r, k=k, n=n: k*r**n for k, n in geometries]

for material in densities:
    for density, function, r in zip(densities[material], functions, radii):
        print(material, density * function(r))
````
